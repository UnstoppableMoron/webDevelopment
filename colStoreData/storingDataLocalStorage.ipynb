# Storing Data with localStorage

### 3 Main Types of Storage

* Cookies
* Session Storage
* Local Storage

```python
var name = prompt("Enter your first name: ");

/* This can be stored as a Key:Value pair in local storage
'name': 'Owen' */
```

```python
/* The JavaScript keyword to store data */
localStorage.setItem("key", "value");
```

```python
/* Retrieve and store data in a variable with Key:Value relationship */

var fName = prompt("Enter your first name: ");

localStorage.setItem("user", fName)

/* To retrieve the data that is stored in local storage */

// var retrieveName = localStorage.getItem("user")
var retrieveName = localStorage.user;
```

### null Data Type
<font color="red">null</font> - an absence of data, no data exists, or data is undefined

The null data type is returned when you use getItem(key) and the data does not exist.

HOWEVER, if you use getItem.key - and the data does not exist, it will return:
<b>undefined</b>

# localStorage Recap

localStorage.setItem("key", "value") ---> key/value pair

Both the key and value parameters MUST BE "strings", so they must be conatined in single or double quotes.

## Saving a List

Save a list of scores:
var grades = [78, 89, 65, 91, 86];

localStorage.setItem("grades", grades);

## Access the first grade

grades[0];

grades[3];

# JSON.stringify() and JSON.parse()

var listOfScores = [20, 45, 10, 344];

localStorage.setItem("scores", JSON.stringify(listOfScores))

In the JSON file, the list will look like "[20, 45, 10, 344]"

To retrieve the data from the JSON file, we use this line of JavaScript:

var resScores = JSON.parse(localStorage.getItem("scores"));

--> [20, 45, 10, 344]

To access items in the list, use the index numbers again:
listOfScores[2] --> the third item in the list

### When to save data in lists:

- collecting multiple iterations of an event
    - [score1, score2, score3, ...]

- multiple items of data to collect
    - [backgroundColor, fontSize, username, password...]
